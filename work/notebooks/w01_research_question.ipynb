## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Research question:**

Can the available search-performance signals help us rank pages by their likelihood of needing a content refresh, so that a content team can decide which pages should be reviewed first?

**Decision:**
Which pages should the content team prioritize for human review?

**Action:**
Review the highest-priority pages and decide whether to refresh, improve, monitor, or leave the content unchanged.

**Cost of a wrong call:**
If a page that needs attention is missed, the team may lose an opportunity to improve its search performance. If a healthy page is incorrectly prioritized, limited content-team time may be spent reviewing a page that does not need a change.


## 3. Quick look at the data (2-3 real numbers)

```python
import pandas as pd

url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

print("Rows:", len(df))
print("Columns:", len(df.columns))
print("\nColumn names:")
print(df.columns.tolist())
```

```python
print("Total pages:", len(df))

print("\nPages by trend direction:")
print(df["trend_direction"].value_counts())

print("\nAverage CTR:", round(df["ctr"].mean(), 4))
print("Average position:", round(df["avg_position"].mean(), 2))
```

The starter dataset contains **30,000 pages** across **44 columns**. In the observed data, **16,262 pages have a “down” trend**, compared with **5,962 stable pages** and **4,388 pages with an “up” trend**. The average CTR is **0.5107**, and the average position is **16.34**.

These observed numbers make the content refresh and decline detection lane worth investigating because a substantial number of pages are marked as declining. The next step is to understand which measurable signals are associated with these observed trends and whether they can help prioritize pages for human review.


## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

### What this work can and cannot claim

This work can identify observed patterns in search-performance data and rank pages that appear more likely to need human review. The analysis can provide directional evidence about which measurable signals are associated with observed declines, such as changes in impressions, clicks, sessions, CTR, and average position.

The output should be treated as decision-support rather than a final decision. A high-priority page is a page that may deserve human review, not a page that is guaranteed to need a content refresh.

This work cannot prove that a particular change in content caused a change in search performance. It also cannot prove that a page will decline in the future or predict exactly how Google will rank a page. Search performance can be affected by many factors that are not available in this dataset.

Therefore, the results will be used to support human review and prioritization, not to make causal claims or claim that the model can predict Google's decisions.
